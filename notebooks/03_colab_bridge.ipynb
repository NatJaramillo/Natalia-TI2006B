# Activity 3: Colab–VS Code Bridge

## Question
How can we use Python to analyze synthetic banking transactions and identify possible fraudulent activity?

## Data provenance
The dataset used in this activity is completely synthetic and was created for educational purposes. It does not contain real customer or banking information.

## Expected output
A simple analysis of banking transactions showing the total number of transactions, the number of legitimate and potentially fraudulent transactions, and their average amounts.

## Objective
To connect the local VS Code notebook with Google Colab, execute the analysis in a clean runtime, and synchronize the final notebook with GitHub.


```python

# Setup: Python version and required packages

import sys
import pandas as pd

print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
print("Environment setup completed successfully!")
```

```python

# Synthetic banking transactions

data = {
    "Transaction_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Amount": [250, 1200, 5000, 150, 8500, 300, 7200, 900],
    "Hour": [10, 14, 3, 12, 2, 16, 1, 11],
    "Distance_km": [2, 5, 150, 3, 200, 4, 180, 8],
    "Status": [
        "Legitimate",
        "Legitimate",
        "Fraud",
        "Legitimate",
        "Fraud",
        "Legitimate",
        "Fraud",
        "Legitimate"
    ]
}

df = pd.DataFrame(data)

print("Synthetic banking transactions:")
display(df)
```

```python

# Transaction analysis

total_transactions = len(df)

status_counts = df["Status"].value_counts()

average_amount = df.groupby("Status")["Amount"].mean()

print("TRANSACTION ANALYSIS")
print("-------------------------")
print("Total transactions:", total_transactions)
print("Legitimate transactions:", status_counts["Legitimate"])
print("Fraudulent transactions:", status_counts["Fraud"])

print("\nAverage transaction amount by status:")
display(average_amount)
```


## Conclusion

The analysis shows that 5 out of 8 transactions were legitimate
and 3 were identified as fraudulent.

The average amount of fraudulent transactions was higher
than the average amount of legitimate transactions.

This simple analysis shows how Python and Pandas can help
organize and understand transaction data.

However, the results are based on synthetic data and do not
represent a real fraud detection model.
